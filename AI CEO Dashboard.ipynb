```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from IPython.display import display, clear_output
import random
import time
from datetime import datetime


# -------------------------------
# AI CEO DATA ENGINE
# -------------------------------

data = {
    "Time": [],
    "Revenue": [],
    "Customers": [],
    "AI_Agents": [],
    "Risk": [],
    "Productivity": []
}


# -------------------------------
# GENERATE BUSINESS DATA
# -------------------------------

def generate_data():

    data["Time"].append(
        datetime.now().strftime("%H:%M:%S")
    )

    data["Revenue"].append(
        random.randint(400000,900000)
    )

    data["Customers"].append(
        random.randint(9000,20000)
    )

    data["AI_Agents"].append(
        random.randint(100,500)
    )

    data["Risk"].append(
        random.randint(5,70)
    )

    data["Productivity"].append(
        random.randint(70,99)
    )



# -------------------------------
# DASHBOARD
# -------------------------------

def show_dashboard():

    clear_output(wait=True)

    generate_data()

    df=pd.DataFrame(data)


    print("="*60)
    print("🚀 AI CEO COMMAND CENTER")
    print("="*60)

    print()

    print(
        f"💰 Revenue: ${df.Revenue.iloc[-1]:,}"
    )

    print(
        f"👥 Customers: {df.Customers.iloc[-1]}"
    )

    print(
        f"🤖 AI Agents Running: {df.AI_Agents.iloc[-1]}"
    )

    print(
        f"⚠️ Risk Score: {df.Risk.iloc[-1]}%"
    )

    print(
        f"⚡ Productivity: {df.Productivity.iloc[-1]}%"
    )


    # Revenue chart

    fig1=go.Figure()

    fig1.add_trace(
        go.Scatter(
            x=df.Time,
            y=df.Revenue,
            mode="lines+markers",
            fill="tozeroy",
            name="Revenue"
        )
    )


    fig1.update_layout(
        title="💰 Revenue Intelligence",
        template="plotly_dark",
        height=400
    )


    display(fig1)



    # AI Agents Chart

    fig2=go.Figure()

    fig2.add_trace(
        go.Bar(
            x=df.Time,
            y=df.AI_Agents,
            name="AI Agents"
        )
    )


    fig2.update_layout(
        title="🤖 AI Workforce Activity",
        template="plotly_dark",
        height=400
    )


    display(fig2)



    # Risk Gauge

    fig3=go.Figure(
        go.Indicator(
            mode="gauge+number",
            value=df.Risk.iloc[-1],
            title={
                "text":"Enterprise Risk"
            },

            gauge={
                "axis":{
                    "range":[0,100]
                }
            }
        )
    )


    fig3.update_layout(
        template="plotly_dark",
        height=350
    )


    display(fig3)



# -------------------------------
# RUN LIVE DASHBOARD
# -------------------------------

for i in range(20):

    show_dashboard()

    time.sleep(3)
```
