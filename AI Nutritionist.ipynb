```python
import gradio as gr


# =====================================
# AI NUTRITION ADVISOR ENGINE
# =====================================

def nutrition_advisor(
    name,
    age,
    gender,
    weight,
    height,
    goal,
    food_type,
    activity
):

    # BMI Calculation

    bmi = weight / ((height / 100) ** 2)


    if bmi < 18.5:
        bmi_status = "Underweight"

        health_advice = """
Your BMI is low.

Advice:
✅ Increase healthy calories
✅ Add protein in every meal
✅ Eat nuts, milk, paneer, eggs
✅ Focus on strength training
"""


    elif bmi < 25:

        bmi_status = "Healthy"

        health_advice = """
Your BMI is in a healthy range.

Advice:
✅ Maintain balanced meals
✅ Continue exercise
✅ Keep good sleep habits
"""


    elif bmi < 30:

        bmi_status = "Overweight"

        health_advice = """
Your BMI is above healthy range.

Advice:
✅ Reduce processed food
✅ Increase vegetables
✅ Control portions
✅ Walk/exercise regularly
"""


    else:

        bmi_status = "High BMI"

        health_advice = """
Focus on gradual lifestyle improvement.

Advice:
✅ Reduce sugary foods
✅ Increase daily activity
✅ Choose high protein meals
"""



    # =====================================
    # MEAL DATABASE
    # =====================================


    veg = {

    "Weight Loss":
"""
🌅 Breakfast:
• Oats with fruits
• Green tea
• Sprouts


☀️ Lunch:
• 2 Roti
• Dal
• Green vegetables
• Salad


🍎 Snacks:
• Apple
• Almonds
• Yogurt


🌙 Dinner:
• Vegetable soup
• Paneer/tofu
• Salad
""",


    "Muscle Gain":
"""
🌅 Breakfast:
• Banana smoothie
• Oats
• Peanut butter


☀️ Lunch:
• Rice
• Dal
• Paneer curry
• Vegetables


🍎 Snacks:
• Nuts
• Milk


🌙 Dinner:
• Roti
• Paneer
• Vegetables
""",


    "Healthy Lifestyle":
"""
🌅 Breakfast:
• Poha/Upma
• Fruits


☀️ Lunch:
• Dal
• Rice
• Vegetables


🍎 Snacks:
• Nuts
• Green tea


🌙 Dinner:
• Roti
• Vegetables
• Salad
"""

}



    nonveg = {


    "Weight Loss":
"""
🌅 Breakfast:
• Boiled eggs
• Oats
• Fruits


☀️ Lunch:
• Grilled chicken
• Brown rice
• Salad


🍎 Snacks:
• Yogurt
• Nuts


🌙 Dinner:
• Fish/chicken soup
• Vegetables
""",


    "Muscle Gain":
"""
🌅 Breakfast:
• Eggs
• Milk
• Banana


☀️ Lunch:
• Chicken breast
• Rice
• Vegetables


🍎 Snacks:
• Protein shake
• Nuts


🌙 Dinner:
• Fish/chicken
• Roti
• Salad
""",



    "Healthy Lifestyle":
"""
🌅 Breakfast:
• Eggs
• Fruits
• Oats


☀️ Lunch:
• Chicken/fish
• Rice
• Vegetables


🍎 Snacks:
• Nuts
• Yogurt


🌙 Dinner:
• Lean protein
• Vegetables
"""
}



    # Select meal plan

    if food_type == "Vegetarian":

        meal = veg[goal]

    else:

        meal = nonveg[goal]



    # Activity Advice

    if activity == "Low":

        exercise = """
Start with:
• 30 minutes walking daily
• Light stretching
"""


    elif activity == "Medium":

        exercise = """
Recommended:
• Strength training 3-4 days/week
• Cardio sessions
"""


    else:

        exercise = """
Maintain:
• Strength training
• Recovery
• Proper protein intake
"""



    # Final Report


    report = f"""

# 🥗 AI Nutrition Advisor Report


Hello {name}! 👋


## Personal Profile

Age:
{age}

Gender:
{gender}

Weight:
{weight} kg

Height:
{height} cm



## Health Analysis

BMI:
{round(bmi,2)}

Status:
{bmi_status}



## 🧠 AI Advice

{health_advice}



## 🎯 Goal

{goal}



## 🍽️ {food_type} Meal Plan


{meal}



## 🏃 Activity Recommendation

{exercise}



## Daily Rules

💧 Drink 2-3 litres water

🥗 Eat vegetables daily

😴 Sleep 7-8 hours

🥚 Include protein source

⚠️ This is general wellness guidance, not medical advice.

"""


    return report



# =====================================
# GRADIO APP UI
# =====================================


app = gr.Interface(

    fn=nutrition_advisor,


    inputs=[


        gr.Textbox(
            label="Name"
        ),


        gr.Number(
            label="Age"
        ),


        gr.Dropdown(
            [
                "Male",
                "Female",
                "Other"
            ],
            label="Gender"
        ),


        gr.Number(
            label="Weight (kg)"
        ),


        gr.Number(
            label="Height (cm)"
        ),


        gr.Dropdown(

            [
                "Weight Loss",
                "Muscle Gain",
                "Healthy Lifestyle"
            ],

            label="Goal"

        ),



        gr.Dropdown(

            [
                "Vegetarian",
                "Non Vegetarian"
            ],

            label="Food Preference"

        ),



        gr.Dropdown(

            [
                "Low",
                "Medium",
                "High"
            ],

            label="Activity Level"

        )

    ],


    outputs=gr.Markdown(),


    title="🥗 AI Nutrition Advisor + Meal Planner",

    description=
    """
    Personalized nutrition assistant
    with veg and non-veg meal suggestions.
    """

)



app.launch()
```
